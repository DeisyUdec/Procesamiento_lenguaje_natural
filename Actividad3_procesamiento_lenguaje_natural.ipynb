```python
#se importa la libreria re para buscar y remplazar caracteres dentro del texto
import re

#se crea la variable texto_prueba para almacenar los caracteres de la frase
texto_prueba = "¡Hola!, hoy es Domingo y voy a comer helado."

#convierte todas la palabras minusculas
texto_normalizado = texto_prueba.lower()
#elimina signos de puntuación y caracteres especiales
texto_normalizado = re.sub(r'[^\w\s]', '', texto_normalizado)
#organiza la frase teniendo en cuenta lo anterior y ajustandola
texto_normalizado = re.sub(r'\s+', ' ', texto_normalizado).strip()

#muestra el resultado
print("Texto sin ajustes: ",texto_prueba)
print("Textos ya ajustada aplicando el metodo normalización de texto: ", texto_normalizado)
```

```python
#Se importa la librería que se utiliza para reconocer la raíz de las palabras
from nltk.stem import SnowballStemmer
# se define el programa que se va atrabaje en español
stemmer = SnowballStemmer("spanish")
#se crea una lista de las palabras que queremos analizar
palabras = ["saltaba", "salto", "saltaría", "comio"]
#se genera el ciclo para realizar cada palabra y determinar su raiz
for palabra in palabras:
    print(palabra, "->", stemmer.stem(palabra))
```

```python
#se importa la librerias que se utilizan para lenguaje natural y para acceder al
# al listado de palabras que se pueden eliminar
import nltk
from nltk.corpus import stopwords
#descarga el listado de palabras que se pueden utilizar
nltk.download('stopwords')
# se carga la frase de prueba
texto = texto_normalizado
#convierte a minusculas y separa las palabras en el arreglo
palabras = texto.lower().split()
#obtiene la lista de palabras comunes en español
stopwords_es = set(stopwords.words('spanish'))
#compara el listado de palabras si esta la elimina
resultado = [palabra for palabra in palabras
             if palabra not in stopwords_es]

print("Texto original", texto)
print("Texto aplicando stopwords",resultado)
```

```python
#importa la herramienta para convertir textos a numeros
from sklearn.feature_extraction.text import TfidfVectorizer
#se crea el arreglo con las frases de prueba
documentos = [
    "el lunes jugamos futbol y comemos pizza",
    "el lunes comemos carne y jugamos parques"
]
#crea el objeto que realiza la transformación en donde determina la frecuencia de las palabras
vectorizador = TfidfVectorizer(use_idf=False)
#aprende acerca de las palabras del arreglo y convierte a representación numerica
tf = vectorizador.fit_transform(documentos)
#muestra las palabras vectorizadas primero la mas comun y la frecuencia de cada una
print(vectorizador.get_feature_names_out())
print(tf.toarray())
```

```python
# se imporata la herramienta que calcula la feceuncia de palabras comunes
from sklearn.feature_extraction.text import TfidfVectorizer
#se crea el arreglo con las frases de prueba
documentos = [
    "el lunes jugamos futbol y comemos pizza",
    "el lunes comemos carne y jugamos parques"
]
#crea el objeto que realiza la transformación en donde determina la frecuencia de las palabras
vectorizador = TfidfVectorizer()
#aprende acerca de las palabras del arreglo y convierte a representación numerica
vectorizador.fit(documentos)
#muestra las palabras y su vectorizadas
print("Palabras:")
print(vectorizador.get_feature_names_out())
print("\nIDF:")
print(vectorizador.idf_)
```

```python
# importa la herramienta de pipeline el cual permite utilizar modelo de IA
from transformers import AutoTokenizer, AutoModelForQuestionAnswering
import torch
from transformers import pipeline
#crea un sistema en deonde se indica que la tarea es responder las preguntas teniendo en cuenta un analisis para obtner respuestas
model_name = "deepset/roberta-base-squad2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)

#ingresa frase de prueba
contexto = "La ganadería es una de las actividades económicas y del sector primario más antiguas e importantes del mundo, fundamentada en la crianza, cuidado y reproducción de animales domésticos para la obtención de recursos esenciales como carne, leche, pieles, lana y tracción animal. Más allá de su valor comercial y de ser el sustento de millones de familias rurales, esta práctica juega un rol crucial en la seguridad alimentaria global al proveer proteínas de alta calidad"
#se genera pregunta prueba
pregunta = "¿A que sector pertenece la ganadería?"
# Procesamos los textos para el modelo
inputs = tokenizer(pregunta, contexto, return_tensors="pt")

# Obtenemos la predicción del modelo
with torch.no_grad():
    outputs = model(**inputs)

#  Buscamos el inicio y fin de la respuesta en el texto
answer_start = torch.argmax(outputs.start_logits)
answer_end = torch.argmax(outputs.end_logits) + 1

# Decodificamos y mostramos el resultado final
respuesta = tokenizer.convert_tokens_to_string(tokenizer.convert_ids_to_tokens(inputs["input_ids"][0][answer_start:answer_end]))
print(respuesta)
```

```python
# importa la herramienta de pipeline el cual permite utilizar modelo de IA
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch
# se crea el sistema de resumir el texto definiendo el modelo
model_name = "csebuetnlp/mT5_multilingual_XLSum"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

#se crea el parrafo de prueba
texto = """
La inteligencia artificial es una tecnología que permite
a los computadores realizar tareas que normalmente requieren
capacidades humanas. Actualmente se utiliza en diferentes
áreas como educación, medicina, industria, transporte y
análisis de información.
"""
# se activa el modelo definiendo la longitud maxima y minima,
#ademas se genera qie el resultado sea determinista
# Procesamos el texto de entrada
inputs = tokenizer(texto, return_tensors="pt", padding=True, truncation=True)

# Generamos el resumen controlando tus parámetros de longitud
with torch.no_grad():
    summary_ids = model.generate(
        inputs["input_ids"],
        max_length=40,
        min_length=15,
        do_sample=False,
        num_beams=4
    )

# Decodificamos el resultado final
resultado = tokenizer.decode(summary_ids, skip_special_tokens=True)
print(resultado)
```

```python
# se importa los modelos convierte las frases a vectores numericos y
#determina la similitud entre los vectores
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
# carga el modelo
modelo = SentenceTransformer("all-MiniLM-L6-v2")
#se crean las frases de prueba
frases = [
    "La inteligencia artificial ayuda a analizar datos.",
    "La IA permite realizar análisis de información."
]
#convierte las frases en representación de vectores
vectores = modelo.encode(frases)
#se compara los vectores
similitud = cosine_similarity(
    [vectores[0]],
    [vectores[1]]
)
#se muestra el resultado
print("Similitud:", similitud[0][0])
```

```python
# importa la herramienta de pipeline el cual permite utilizar modelo de IA
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch
#se crea el modelo de traducción
model_name = "Helsinki-NLP/opus-mt-es-en"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

#se ingresa frase de prueba
texto = "La inteligencia artificial está cambiando el mundo."
# el modelo analiza y traduce
inputs = tokenizer(texto, return_tensors="pt", padding=True, truncation=True)
with torch.no_grad():
    translated_tokens = model.generate(**inputs)
resultado = tokenizer.decode(translated_tokens[0], skip_special_tokens=True)


#muestra resultados
print(resultado)
```

```python
#importa el contador para determinar la cantidad de veces que esta una palabra
from collections import Counter
#importa la herramienta para trabajar con expresiones reguares
import re
# se crea el arreglo con las frases de prueba
documentos = [
    "La inteligencia artificial analiza datos",
    "La inteligencia artificial ayuda a tomar decisiones",
    "El análisis de datos utiliza inteligencia artificial"
]
#une todos los documentos en uno solo
texto = " ".join(documentos).lower()
#identica y extrae palabras del texto
palabras = re.findall(r'\b\w+\b', texto)
#selecciona las principales palbras que aparcen con mayor frecuncia
frecuencias = Counter(palabras)
#mustra los resultados
print(frecuencias.most_common(10))
```
